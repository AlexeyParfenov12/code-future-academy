# Урок 1.14. Разметка данных для задач ИИ
Практикум со stdout-заданиями.

```
import pandas as pd

MESSAGES_FILE = 'messages_for_labeling_1_14.csv'
ANNOTATIONS_FILE = 'annotations_by_students_1_14.csv'
IMAGE_PAIRS_FILE = 'image_pairs_for_labeling_1_14.csv'
```

## Задание 1. Размер таблицы сообщений
Ожидаемый вывод: `12 5`

```
df = pd.read_csv(MESSAGES_FILE)
print(df.shape[0], df.shape[1])
```

## Задание 2. Пустые метки
Ожидаемый вывод: `12`

```
empty_labels = df['label'].isna().sum() + (df['label'].astype(str).str.strip() == '').sum()
print(empty_labels)
```

## Задание 3. Разметка по правилу
Ожидаемый вывод: `needs_review=2;normal=5;spam=5`

```
def make_label(row):
    if row['has_link'] == 'yes' and row['has_money_or_prize_word'] == 'yes':
        return 'spam'
    if row['has_link'] == 'no' and row['has_money_or_prize_word'] == 'no':
        return 'normal'
    return 'needs_review'

labeled = df.copy()
labeled['label'] = labeled.apply(make_label, axis=1)
counts = labeled['label'].value_counts().sort_index()
print(';'.join(f'{name}={count}' for name, count in counts.items()))
```

## Задание 4. Спорные сообщения
Ожидаемый вывод: `5,9`

```
ids = labeled.loc[labeled['label'] == 'needs_review', 'id'].tolist()
print(','.join(map(str, ids)))
```

## Задание 5. Согласованность разметчиков
Ожидаемый вывод: `8`

```
ann = pd.read_csv(ANNOTATIONS_FILE)
agree = (ann['annotator_1'] == ann['annotator_2']) & (ann['annotator_1'] == ann['annotator_3'])
print(agree.sum())
```

## Задание 6. Разногласия разметчиков
Ожидаемый вывод: `5,9`

```
disagree_ids = ann.loc[~agree, 'id'].tolist()
print(','.join(map(str, disagree_ids)))
```

## Задание 7. Пары изображений
Ожидаемый вывод: `B,A,A,A`

```
# Здесь ответы заданы по смыслу описаний.
answers = ['B', 'A', 'A', 'A']
print(','.join(answers))
```

## Задание 8. Исправьте код, предложенный ИИ
Ниже намеренно оставлена корректная версия. На платформе может быть дан ошибочный фрагмент.

```
# Корректная функция учитывает оба признака и не угадывает спорные случаи.
def make_label_fixed(row):
    if row['has_link'] == 'yes' and row['has_money_or_prize_word'] == 'yes':
        return 'spam'
    if row['has_link'] == 'no' and row['has_money_or_prize_word'] == 'no':
        return 'normal'
    return 'needs_review'

fixed = df.copy()
fixed['label'] = fixed.apply(make_label_fixed, axis=1)
counts = fixed['label'].value_counts().sort_index()
print(';'.join(f'{name}={count}' for name, count in counts.items()))
```
