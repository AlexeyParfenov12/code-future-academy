## 1. Импорт библиотек


```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## 2. Загружаем датасет
В таблице нет правильного класса. Есть только персонажи и их признаки.


```
df = pd.read_csv('game_characters_sklearn.csv')
df.head()
```

```
df.info()
```

## 3. Карта признаков без кластеров
Сначала смотрим на объекты как на точки.


```
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='speed', y='power', s=90)
for _, row in df.iterrows():
    plt.text(row['speed'] + 0.08, row['power'] + 0.08, row['character'], fontsize=8)
plt.title('Персонажи на карте признаков')
plt.xlabel('speed')
plt.ylabel('power')
plt.grid(True)
plt.show()
```

## 4. Выбираем признаки для кластеризации
В кластеризации нет `y`: алгоритм не получает правильные группы заранее.


```
X = df[['speed', 'power']]
X.head()
```

## 5. Запускаем KMeans
Попросим алгоритм найти 3 группы.


```
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df['cluster'] = model.fit_predict(X)
df.head(10)
```

## 6. Визуализируем найденные группы
Цвет показывает номер кластера. Номер сам по себе ещё не является названием группы.


```
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='speed', y='power', hue='cluster', palette='Set2', s=100)
for _, row in df.iterrows():
    plt.text(row['speed'] + 0.08, row['power'] + 0.08, row['character'], fontsize=8)
plt.title('Кластеры персонажей')
plt.xlabel('speed')
plt.ylabel('power')
plt.grid(True)
plt.show()
```

## 7. Интерпретируем кластеры
Посмотрим средние значения признаков внутри каждого кластера.


```
cluster_profile = df.groupby('cluster')[['speed', 'power', 'defense']].mean().round(1)
cluster_profile
```

Запишите смысл каждого кластера своими словами. Например: быстрые персонажи, сильные персонажи, сбалансированные персонажи.


## 8. Что изменится при другом числе кластеров?


```
for k in [2, 3, 4]:
    model_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    df[f'cluster_{k}'] = model_k.fit_predict(X)

df[['character', 'speed', 'power', 'cluster_2', 'cluster_3', 'cluster_4']].head(12)
```

```
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='speed', y='power', hue='cluster_4', palette='Set2', s=100)
plt.title('Пример: 4 кластера')
plt.xlabel('speed')
plt.ylabel('power')
plt.grid(True)
plt.show()
```
