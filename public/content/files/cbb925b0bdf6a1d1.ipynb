На этом занятии мы берём реальный табличный датасет Iris, готовим признаки, обучаем простую модель классификации и получаем предсказания. Метрики качества пока не считаем: они будут отдельно на следующем уроке.


## Что понадобится
- `iris_real_dataset.csv` — таблица с измерениями цветков ириса;
- `iris_new_examples.csv` — несколько новых объектов для предсказания;
- `pandas`;
- `scikit-learn`.


```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

## Шаг 1. Загружаем датасет
Сначала читаем CSV-файл и смотрим первые строки.


```
df = pd.read_csv('iris_real_dataset.csv')
df.head()
```

## Шаг 2. Изучаем структуру таблицы
Важно увидеть количество строк и столбцов, названия столбцов, типы данных и возможные пропуски.


```
print("Размер таблицы:", df.shape)
print("Столбцы:", list(df.columns))
df.info()
```

```
df.describe()
```

```
df.isna().sum()
```

## Шаг 3. Готовим признаки и целевую переменную
`X` — признаки цветка, `y` — правильный вид цветка.


```
df = df.dropna()

X = df[[
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]]

y = df["species"]

display(X.head())
display(y.head())
```

## Шаг 4. Делим данные на обучающую и тестовую части
Модель учится на `X_train` и `y_train`, а предсказания затем смотрим на новых для неё строках `X_test`.


```
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

## Шаг 5. Обучаем дерево решений
Дерево решений удобно для первого практикума: его можно представить как последовательность вопросов по признакам.


```
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
print("Модель обучена.")
```

## Шаг 6. Получаем предсказания на тестовых строках
Пока не считаем отдельные метрики. Просто смотрим, какие ответы модель дала новым для неё объектам.


```
predictions = model.predict(X_test)

result = X_test.copy()
result["true_species"] = y_test.values
result["predicted_species"] = predictions

result.head(10)
```

## Шаг 7. Предсказания для новых объектов
Теперь загрузим отдельный файл с новыми цветками. У них должны быть те же признаки, что и в `X`.


```
new_examples = pd.read_csv('iris_new_examples.csv')
new_examples
```

```
new_predictions = model.predict(new_examples)

new_result = new_examples.copy()
new_result["predicted_species"] = new_predictions
new_result
```
