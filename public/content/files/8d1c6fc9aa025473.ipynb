## Идея урока
Регрессия предсказывает число. Поэтому мы сравниваем два числа: фактическое значение и прогноз модели.

Абсолютная ошибка показывает, на сколько единиц модель промахнулась.


```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error
```

## Шаг 1. Загружаем таблицу с прогнозами


```
df = pd.read_csv('regression_predictions_basic.csv')
df.head()
```

## Шаг 2. Считаем абсолютную ошибку
`abs()` убирает знак минус. Нам важно не направление промаха, а расстояние между фактом и прогнозом.


```
df['error'] = abs(df['real_score'] - df['predicted_score'])
df
```

## Шаг 3. Ищем самый точный прогноз и самый большой промах


```
best_row = df.loc[df['error'].idxmin()]
worst_row = df.loc[df['error'].idxmax()]
print('Самый точный прогноз:')
display(best_row)
print('Самый большой промах:')
display(worst_row)
```

## Шаг 4. Считаем среднюю абсолютную ошибку


```
mean_error = df['error'].mean()
print('Средняя ошибка:', mean_error)
print('В среднем модель промахивается примерно на', round(mean_error, 1), 'очков')
```

## Шаг 5. Проверяем через готовую функцию sklearn


```
mae = mean_absolute_error(df['real_score'], df['predicted_score'])
print('MAE:', mae)
```

## Шаг 6. Строим визуальное сравнение факта и прогноза


```
plt.figure(figsize=(8, 4))
plt.plot(df['player_id'], df['real_score'], marker='o', label='Факт')
plt.plot(df['player_id'], df['predicted_score'], marker='o', label='Прогноз')
plt.xlabel('Игрок')
plt.ylabel('Очки')
plt.title('Фактические и предсказанные очки')
plt.legend()
plt.grid(True)
plt.show()
```

## Уровень 3. Сравнение двух моделей
Посчитаем среднюю ошибку для двух вариантов прогнозов.


```
two = pd.read_csv('regression_predictions_two_models.csv')
two['error_A'] = abs(two['real_score'] - two['model_A_prediction'])
two['error_B'] = abs(two['real_score'] - two['model_B_prediction'])
display(two.head())
print('Средняя ошибка модели A:', two['error_A'].mean())
print('Средняя ошибка модели B:', two['error_B'].mean())
```
