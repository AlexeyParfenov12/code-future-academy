В этом ноутбуке модель предсказывает числовой результат игрока по времени тренировки и прошлому результату.


## 1. Импорт библиотек


```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

## 2. Загружаем данные


```
df = pd.read_csv('game_scores_regression.csv')
df.head()
```

```
df.info()
```

## 3. Смотрим связь одного признака и результата
Этот график нужен для понимания: чем больше время тренировки, тем выше обычно результат, хотя точки не лежат идеально на одной линии.


```
plt.scatter(df['training_minutes'], df['score'])
plt.xlabel('training_minutes')
plt.ylabel('score')
plt.title('Время тренировки и результат')
plt.show()
```

## 4. Готовим признаки и целевой результат
`X` — признаки, `y` — число, которое модель должна предсказать.


```
X = df[['training_minutes', 'previous_score']]
y = df['score']
display(X.head())
display(y.head())
```

## 5. Делим данные на train и test


```
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
print('X_train:', X_train.shape)
print('X_test:', X_test.shape)
```

## 6. Обучаем LinearRegression


```
model = LinearRegression()
model.fit(X_train, y_train)
print('Модель обучена')
```

## 7. Получаем предсказания и считаем ошибку


```
predictions = model.predict(X_test)
result = X_test.copy()
result['real_score'] = y_test.values
result['predicted_score'] = predictions.round(1)
result['error'] = abs(result['real_score'] - result['predicted_score'])
result.head(10)
```

```
mean_error = result['error'].mean()
print('Средняя ошибка:', round(mean_error, 2))
```

## 8. Проверяем через готовую функцию sklearn


```
mae = mean_absolute_error(y_test, predictions)
print('MAE:', round(mae, 2))
```

## 9. Прогноз для новых игроков


```
new_players = pd.read_csv('game_scores_new_players.csv')
new_X = new_players[['training_minutes', 'previous_score']]
new_players['predicted_score'] = model.predict(new_X).round(1)
new_players
```
