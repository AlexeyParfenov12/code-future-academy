# Урок 3.7. Оценка качества классификации: accuracy
Практикум для 10-11 классов и СПО. Для 8-9 классов основная работа выполняется по таблицам вручную.

На уроке мы сравниваем правильные ответы с предсказаниями модели и считаем долю верных ответов.


## 1. Сначала повторяем ручную логику в Python
Формула: `accuracy = количество верных ответов / количество всех ответов`.


```
import pandas as pd
```

```
df = pd.read_csv('classification_results_level_1.csv')
df
```

```
df['is_correct'] = df['true_label'] == df['predicted_label']
df
```

```
correct_count = df['is_correct'].sum()
total_count = len(df)
accuracy = correct_count / total_count
print('Верных ответов:', correct_count)
print('Всего примеров:', total_count)
print('Accuracy:', accuracy)
print('Accuracy в процентах:', round(accuracy * 100, 1))
```

## 2. Готовая функция accuracy_score
Теперь проверим тот же результат через функцию из Scikit-learn.


```
from sklearn.metrics import accuracy_score
accuracy_from_sklearn = accuracy_score(df['true_label'], df['predicted_label'])
print(accuracy_from_sklearn)
```

## 3. Сравнение двух моделей
Здесь есть правильный класс и предсказания двух моделей. Нужно понять, какая модель дала больше верных ответов на этой таблице.


```
two = pd.read_csv('classification_results_two_models.csv')
two.head()
```

```
acc_a = accuracy_score(two['true_label'], two['model_A_prediction'])
acc_b = accuracy_score(two['true_label'], two['model_B_prediction'])
print('Accuracy модели A:', round(acc_a, 3), '=', round(acc_a * 100, 1), '%')
print('Accuracy модели B:', round(acc_b, 3), '=', round(acc_b * 100, 1), '%')
```

## 4. Accuracy после обучения модели на Iris
Этот блок связан с предыдущим уроком: обучаем дерево решений и считаем долю верных ответов на тестовой части.


```
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

iris = pd.read_csv('iris_accuracy_dataset.csv')
X = iris[['sepal_length_cm', 'sepal_width_cm', 'petal_length_cm', 'petal_width_cm']]
y = iris['species']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)
print('Accuracy дерева решений:', round(accuracy, 3), '=', round(accuracy * 100, 1), '%')
```

## 5. Расширение для уровня 3
Попробуйте изменить глубину дерева и сравнить accuracy. Важно: вывод делаем аккуратно — только по этой тестовой части.


```
for depth in [1, 2, 3, 4, 5]:
    model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    acc = accuracy_score(y_test, predictions)
    print('max_depth =', depth, '| accuracy =', round(acc, 3), '|', round(acc * 100, 1), '%')
```
