На этом занятии мы не строим алгоритм с нуля. Мы берём готовую модель из библиотеки **Scikit-learn**, передаём ей признаки и правильные ответы, а затем получаем первые предсказания для новых объектов.


## Что понадобится
- файл `messages_train.csv`;
- библиотека `pandas`;
- библиотека `scikit-learn`.

В датасете каждая строка — это одно сообщение, а столбец `label` — правильный класс (`spam`, `normal`, `unclear`).


```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

## Шаг 1. Загружаем таблицу
Сначала читаем CSV-файл и смотрим первые строки.


```
df = pd.read_csv('messages_train.csv')
df.head()
```

## Шаг 2. Отделяем признаки и целевую переменную
`X` — признаки, `y` — правильные ответы.


```
X = df[['text_len', 'has_link', 'has_discount', 'has_urgent']]
y = df['label']
print('Признаки:')
display(X.head())
print('Целевая переменная:')
display(y.head())
```

## Шаг 3. Делим данные на обучающую и тестовую части
На обучающей части модель находит закономерности, на тестовой — проверяется на новых примерах.


```
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
print('Размер X_train:', X_train.shape)
print('Размер X_test:', X_test.shape)
```

## Шаг 4. Обучаем модель
Для первого запуска подойдёт `DecisionTreeClassifier` — модель дерева решений. Она по очереди проверяет признаки и двигается по ветвям к итоговому классу.


```
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
print('Модель обучена.')
```

## Шаг 5. Получаем первые предсказания
Теперь передадим модели несколько новых сообщений в виде таблицы с теми же признаками.


```
new_examples = pd.read_csv('messages_new_examples.csv')
new_examples
```

```
predictions = model.predict(new_examples)
result = new_examples.copy()
result['predicted_label'] = predictions
result
```

## Шаг 6. Пробуем проверить модель на тестовой части
Здесь мы пока не считаем отдельные метрики подробно. Мы просто сравниваем, как модель предсказала несколько новых для неё примеров.


```
test_result = X_test.copy()
test_result['true_label'] = y_test.values
test_result['predicted_label'] = model.predict(X_test)
test_result.head(10)
```
