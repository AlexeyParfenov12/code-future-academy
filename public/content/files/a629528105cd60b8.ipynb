## Задание 4. Выбор класса моделью (уровень 2)

Откройте файл `model_probabilities.csv`. Для каждой строки выберите класс с максимальной вероятностью.

```
probs = pd.read_csv('model_probabilities.csv')
probs
```

```
prob_columns = ['p_spam', 'p_normal', 'p_promo']
class_names = {'p_spam': 'spam', 'p_normal': 'normal', 'p_promo': 'promo'}

probs['max_probability_column'] = probs[prob_columns].idxmax(axis=1)
probs['predicted_class'] = probs['max_probability_column'].map(class_names)
probs[['example_id', 'text_short', 'p_spam', 'p_normal', 'p_promo', 'predicted_class', 'true_class']]
```

## Задание 5. Слабая уверенность модели (уровень 2)

Найдите строки, где самая большая вероятность отличается от второй меньше чем на 0.15.

```
sorted_probs = probs[prob_columns].apply(lambda row: sorted(row, reverse=True), axis=1)
probs['confidence_gap'] = sorted_probs.apply(lambda values: values[0] - values[1])
weak_confidence = probs[probs['confidence_gap'] < 0.15]
weak_confidence[['example_id', 'text_short', 'p_spam', 'p_normal', 'p_promo', 'confidence_gap', 'predicted_class']]
```

## Задание 6. Проверка ответа модели (уровень 2)

Сравните `predicted_class` с `true_class` и посчитайте долю верных ответов.

```
probs['is_correct'] = probs['predicted_class'] == probs['true_class']
correct_count = probs['is_correct'].sum()
accuracy = correct_count / len(probs)
print('Верных ответов:', correct_count)
print('Всего примеров:', len(probs))
print('Доля верных ответов:', accuracy)
probs[['example_id', 'predicted_class', 'true_class', 'is_correct']]
```
