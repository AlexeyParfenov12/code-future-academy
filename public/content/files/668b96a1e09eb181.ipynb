## 0. Подключаем библиотеки

```
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", palette="Set2")
```

## Уровень 1. По образцу: график по спискам

**Задания 1–3.** Постройте график по спискам, найдите максимум и минимум, сохраните PNG-файл.

```
days = ["Пн", "Вт", "Ср", "Чт", "Пт"]
screen_time = [3.5, 4.0, 6.0, 5.5, 7.0]

plt.figure(figsize=(7, 4))
sns.barplot(x=days, y=screen_time, color="skyblue")
plt.title("Экранное время по дням")
plt.xlabel("День недели")
plt.ylabel("Экранное время, часы")
plt.tight_layout()
plt.savefig("l1_screen_time_barplot.png", dpi=150)
plt.show()

max_index = screen_time.index(max(screen_time))
min_index = screen_time.index(min(screen_time))

print("L1_MAX_DAY=", days[max_index], sep="")
print("L1_MIN_DAY=", days[min_index], sep="")
print("L1_FILE_CREATED=", os.path.exists("l1_screen_time_barplot.png"), sep="")
```

Напишите вывод в 1–2 предложениях. Не надо доказывать причину, достаточно описать, что видно.

```
l1_conclusion = "На диаграмме видно, что максимальное экранное время было в пятницу, а минимальное — в понедельник."
print("L1_CONCLUSION=", l1_conclusion, sep="")
```

## Уровень 2. В изменённых условиях: CSV, DataFrame и hue

**Задания 4–7.** Откройте CSV, проверьте размер таблицы, постройте графики с цветовой группировкой.

```
df = pd.read_csv("students_visualization_1_17_10_11_spo.csv")

print("L2_ROWS=", df.shape[0], sep="")
print("L2_COLUMNS=", list(df.columns), sep="")
df.head()
```

```
# Задание 5: barplot по уровню обучения и экранному времени.
# Можно заменить hue_column на "stress_level", если хотите сравнить уровни стресса.
hue_column = "uses_ai_for_study"

plt.figure(figsize=(8, 4.5))
sns.barplot(
    data=df,
    x="grade_level",
    y="screen_time_hours",
    hue=hue_column,
    errorbar=None
)
plt.title("Среднее экранное время по уровню обучения")
plt.xlabel("Уровень обучения")
plt.ylabel("Экранное время, часы")
plt.legend(title="ИИ для учёбы")
plt.tight_layout()
plt.show()

print("L2_HUE_USED=", hue_column, sep="")
```

```
# Задание 7: countplot по подготовке к экзамену.
plt.figure(figsize=(8, 4.5))
sns.countplot(
    data=df,
    x="exam_preparation",
    hue="group"
)
plt.title("Количество обучающихся по уровню подготовки")
plt.xlabel("Уровень подготовки")
plt.ylabel("Количество")
plt.legend(title="Группа")
plt.tight_layout()
plt.savefig("l2_exam_preparation_countplot.png", dpi=150)
plt.show()

print("L2_COUNT_FILE_CREATED=", os.path.exists("l2_exam_preparation_countplot.png"), sep="")
```

```
# Короткая сводка для вывода.
summary = df.groupby("grade_level")["screen_time_hours"].mean().round(2)
print("L2_MEAN_SCREEN_BY_GRADE=", summary.to_dict(), sep="")

l2_conclusion = "Цвет помогает сравнить не только уровень обучения, но и отдельные группы внутри него."
print("L2_CONCLUSION=", l2_conclusion, sep="")
```

## Уровень 3. Самостоятельная настройка: связь признаков и мини-отчёт

**Задания 8–10.** Постройте scatterplot и lineplot, затем создайте мини-отчёт.

```
plt.figure(figsize=(8, 4.8))
sns.scatterplot(
    data=df,
    x="screen_time_hours",
    y="sleep_hours",
    hue="stress_level",
    size="study_hours",
    sizes=(40, 180)
)
plt.title("Экранное время, сон и стресс")
plt.xlabel("Экранное время, часы")
plt.ylabel("Сон, часы")
plt.tight_layout()
plt.savefig("l3_scatter_screen_sleep.png", dpi=150)
plt.show()

print("L3_SCATTER_FILE_CREATED=", os.path.exists("l3_scatter_screen_sleep.png"), sep="")
```

```
week_df = pd.read_csv("week_visualization_1_17_10_11_spo.csv")

plt.figure(figsize=(8, 4.5))
sns.lineplot(
    data=week_df,
    x="day",
    y="study_hours",
    hue="group",
    marker="o"
)
plt.title("Изменение времени подготовки по дням")
plt.xlabel("День недели")
plt.ylabel("Часы подготовки")
plt.tight_layout()
plt.savefig("l3_week_lineplot.png", dpi=150)
plt.show()

print("L3_WEEK_FILE_CREATED=", os.path.exists("l3_week_lineplot.png"), sep="")
```
