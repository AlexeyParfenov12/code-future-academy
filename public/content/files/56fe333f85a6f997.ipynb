## 1. Выбор датасета и вопроса

```
dataset_name = 'students_lifestyle_mini.csv'  # замените файл при необходимости
question = 'Как различается подготовка к экзаменам в разных группах?'
print('Датасет:', dataset_name)
print('Вопрос:', question)
```

```
df = pd.read_csv(dataset_name)
display(df.head())
print('Размер таблицы:', df.shape)
print('Столбцы:', list(df.columns))
```

```
object_description = 'один ученик или студент в таблице'
important_features = ['grade_level', 'screen_time_hours', 'sleep_hours', 'exam_preparation']
print('Объект анализа:', object_description)
print('Важные признаки:', important_features)
```

```
df.info()
display(df.describe(include='all'))
print('Пропуски по столбцам:')
print(df.isna().sum())
print('Дубли строк:', df.duplicated().sum())
```

## 9. Задание с юнит-тестами

```
def check_mini_analysis(report: dict) -> list[str]:
    issues = []

    dataset = report.get('dataset', '')
    obj = report.get('object', '')
    features = report.get('features', [])
    metrics = report.get('metrics', [])
    visualizations = report.get('visualizations', [])
    conclusion = report.get('conclusion', '')

    if not isinstance(dataset, str) or not dataset.strip():
        issues.append('MISSING_DATASET')
    if not isinstance(obj, str) or not obj.strip():
        issues.append('MISSING_OBJECT')

    clean_features = [x for x in features if isinstance(x, str) and x.strip()] if isinstance(features, list) else []
    if len(clean_features) < 3:
        issues.append('TOO_FEW_FEATURES')

    clean_metrics = [x for x in metrics if isinstance(x, str) and x.strip()] if isinstance(metrics, list) else []
    if len(clean_metrics) < 2:
        issues.append('TOO_FEW_METRICS')

    clean_visualizations = [x for x in visualizations if isinstance(x, str) and x.strip()] if isinstance(visualizations, list) else []
    if len(clean_visualizations) < 1:
        issues.append('NO_VISUALIZATION')

    if not isinstance(conclusion, str) or not conclusion.strip():
        issues.append('EMPTY_CONCLUSION')
    elif len(conclusion.strip()) < 20:
        issues.append('TOO_SHORT_CONCLUSION')

    return issues if issues else ['OK']
```

```
valid_report = {
    'dataset': 'students_lifestyle_mini.csv',
    'object': 'ученик или студент',
    'features': ['grade_level', 'screen_time_hours', 'sleep_hours'],
    'metrics': ['mean_screen_time', 'count_exam_preparation'],
    'visualizations': ['countplot'],
    'conclusion': 'В выбранном датасете видно распределение по подготовке к экзаменам.'
}
assert check_mini_analysis(valid_report) == ['OK']
assert check_mini_analysis({})[:3] == ['MISSING_DATASET', 'MISSING_OBJECT', 'TOO_FEW_FEATURES']
print('Открытые тесты пройдены')
```
