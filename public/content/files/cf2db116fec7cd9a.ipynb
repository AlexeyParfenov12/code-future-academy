## 1. Открываем основной CSV

```
df = pd.read_csv("student_lifestyle_1_15_10_11_spo.csv")
df.head()
```

## 2. Размер таблицы и столбцы

```
print(df.shape[0], df.shape[1])
print(len(df))
print(",".join(df.columns))
```

## 3. Фильтрация по условиям

```
print(len(df[df["country"] == "Japan"]))
print(len(df[df["uses_ai_for_study"] == "Yes"]))
print(len(df[df["stress_level"] >= 8]))
print(len(df[(df["stress_level"] >= 8) & (df["sleep_hours"] < 6)]))
```

## 4. CSV с другим разделителем и кодировкой
